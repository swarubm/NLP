```python
mamatha_reviews = []
```

```python
pattern_to_find = 'mamatha'
```

```python
for reviews in customer_reviews:
    if (re.search(pattern_to_find,reviews)):
        mamatha_reviews.append(reviews)
```

```python
print(mamatha_reviews)
```
