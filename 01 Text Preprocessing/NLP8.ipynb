```python
data['review_lowercase'] = data['Review'].str.lower()
```

```python
nltk.download('stopwords')
en_stopwords = stopwords.words('english')
en_stopwords.remove("not")
```

```python
data['review_no_stopwords'] = data['review_lowercase'] \
                                .apply(lambda x: ' ' \
                                .join([word for word \
                                in x.split() if word \
                                not in (en_stopwords)]))
```

```python
data['review_no_stopwords_no_punct'] = data \
                                    .apply(lambda x: \
                                    re.sub(r"[*]", \
                                    "star", \
                                    x['review_no_stopwords'] \
                                    ), axis=1)
```

# Tokenizing

```python
nltk.download('punkt')
nltk.download('punkt_tab')
data['tokenized'] = data.apply(lambda x: \
                               word_tokenize( \
                               x['review_no_stopwords_no_punct'] \
                               ), axis=1)
```

```python
data['tokenized'][1]
```

```python
##Stemming
```

```python
ps = PorterStemmer()
data["stemmed"] = data["tokenized"] \
                  .apply(lambda tokens: \
                  [ps.stem(token) \
                   for token in tokens])
```

```python
data.head()
```

```python
data['stemmed'][1]
```

```python
##Lemmatization
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
nltk.download('wordnet')
data["lemmatized"] = data["tokenized"] \
                    .apply(lambda tokens: \
                    [lemmatizer.lemmatize(token) \
                     for token in tokens])
```

```python
data['lemmatized'][1]
```

```python
data.head()
```

```python
##N-grams
```

```python
tokens_clean = sum(data['lemmatized'], [])
# unigrams: n=1
unigrams = (pd.Series \
            (nltk.ngrams(tokens_clean, 1)) \
            .value_counts())
print(unigrams)
```

```python
# bigrams: n=2
bigrams = (pd.Series \
           (nltk.ngrams(tokens_clean, 2)) \
           .value_counts())
print(bigrams)
```

```python
ngrams_4 = (pd.Series \
            (nltk.ngrams(tokens_clean, 4)) \
            .value_counts())
print(ngrams_4)
```
