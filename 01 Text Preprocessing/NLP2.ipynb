```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
en_stopwords = stopwords.words('english')
```

```python
print(en_stopwords)
```

```python
sentences = "An AI engineer is a specialized software professional who designs, develops, and deploys artificial intelligence systems, combining machine learning, data science, and software engineering to solve complex problems across industries. They bridge the gap between cutting‑edge AI research and real‑world applications, ensuring models are scalable, reliable, and ethically implemented."
```

```python
sentance_no_stopwords = ' '.join([word for word in sentences.split() if word not in (en_stopwords)])
print(sentance_no_stopwords)
```

```python
en_stopwords.remove("some")
en_stopwords.remove("ourselves")
```
