## stemming

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
tokens_to_stem = ['connecting', 'connected', 'connectivity', 'connect', 'connects','connecting']
```

```python
for tokens in tokens_to_stem:
    print(tokens,":",ps.stem(tokens))
```

```python
learn_tokens = ['learned', 'learning', 'learn', 'learns', 'learner', 'learners']
```

```python
for tokens in learn_tokens:
    print(tokens, "-->",ps.stem(tokens))
```
