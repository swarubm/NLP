```python
import nltk
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
tokens = [
    'Natural', 'Language', 'Processing', 'is', 'fun',
    'Machine', 'Learning', 'models', 'analyze', 'data',
    'Python', 'code', 'runs', 'in', 'notebooks',
    'Stopwords', 'removal', 'cleans', 'text', 'inputs',
    'Translation', 'pipelines', 'convert', 'German', 'English',
    'SpaCy', 'handles', 'entities', 'fast', 'efficient',
    'TextBlob', 'simplifies', 'sentiment', 'analysis', 'tasks',
    'VADER', 'detects', 'emoji', 'tone', 'tweets',
    'Gensim', 'discovers', 'topics', 'using', 'Word2Vec','to','detects','language','to','tasks','to','data'
]
```

```python
print(tokens)
```

```python
unigrams = (pd.Series(nltk.ngrams(tokens,1)))
```

```python
print(unigrams)
```

```python
(pd.Series(nltk.ngrams(tokens,1))).value_counts()
```

```python
print(unigrams[:5])
```

```python
unigrams[:5].sort_values().plot.bar(color='lightsalmon',width = 9,figsize=(12,8))
plt.title("5 most occured unigrams")
```

```python
unigram_series = pd.Series(unigrams).value_counts()
unigram_series.head(5).plot.barh(color='lightsalmon', width=.9, figsize=(12, 8))
plt.title('10 Most Frequently Occurring Unigrams')
plt.show()
```

```python
bigram = (pd.Series(nltk.ngrams(tokens,2))).value_counts()
print(bigram)
```
