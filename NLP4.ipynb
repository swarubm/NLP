## Tokenization
 breaking text into smaller units(tokens)

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python
sentences = "yashas is the first name. lishas is the second son name"
```

```python
sent_tokenize(sentences)
```

```python
sentence = "her name is suma"
word_tokenize(sentence)
```

```python
sentences = "suma is a teacher having a husband rudresh and she has two sons. one is yashas. second one is lishu"
word_tokenize(sentences)
```
